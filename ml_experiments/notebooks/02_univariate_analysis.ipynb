# Sub-session S011: Univariate Analysis
In this notebook, we visualize individual features to understand their distributions and spot outliers.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Set a professional visual theme
sns.set_theme(style="whitegrid")

# Load data
df = pd.read_csv('../data/raw/crop_recommendation.csv')
numeric_features = ['N', 'P', 'K', 'temperature', 'humidity', 'ph', 'rainfall']
print("Data loaded. Ready to plot!")
```

## 1. Histograms (Distribution Shapes)
Look at the curves. Are they symmetrical (Normal)? Or are they leaning to one side (Skewed)?

```python
plt.figure(figsize=(15, 10))
for i, feature in enumerate(numeric_features, 1):
    plt.subplot(3, 3, i)
    sns.histplot(df[feature], kde=True, color='skyblue')
    plt.title(f'Distribution of {feature}')
plt.tight_layout()
plt.show()
```

## 2. Boxplots (Outlier Detection)
Look for black dots outside the 'whiskers'. These represent extreme values.

```python
plt.figure(figsize=(15, 10))
for i, feature in enumerate(numeric_features, 1):
    plt.subplot(3, 3, i)
    sns.boxplot(y=df[feature], color='lightgreen')
    plt.title(f'Boxplot of {feature}')
plt.tight_layout()
plt.show()
```
